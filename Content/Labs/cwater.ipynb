# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

De fysische natuurkunde leert ons dat temperatuur een verband heeft met energie; hoe hoger de temperatuur van een stof, hoe meer energie of 'warmte' de stof heeft. Om een stof op te warmen, moet iets dus energie af staan aan die stof. En andersom, als een stof afkoelt, geeft de stof energie af aan zijn omgeving. Bij sommige stoffen kost het meer energie om te warmen dan bij andere stoffen. De hoeveelheid energie die het per kilogram per graad opwarming kost, heet de soortelijke warmte. Metalen hebben bijvoorbeeld een relatief lage soortelijke warmte en water een relatief hoge. In dit experiment wordt de soortelijke warmte van water bepaald. Daarmee wordt gebruikt gemaakt van de formule voor warmte:


${Q = c \Delta Tm}$   

Hierin is Q de warmte die de stof krijgt, c de soortelijke warmte, $\Delta$ T het temperatuurverschil en m de massa van de stof. Dit is een lineair verband en in dit experiment wordt deze formule gebruikt om mee te fitten. De helling die hierbij bevonden wordt kan worden gebruikt om c te bepalen.

## Data analyse
Eerst wordt de totale warmte berekend die is toegevoegd aan het water. Dat wordt gedaan door middel van E = P * t.

Daarna wordt er een lineaire fit gedaan om te laten zien dat er daadwerkelijk sprake is van een lineair verband.

De berekende waarde van Q wordt gelijkgesteld aan $cm\Delta T$. Deze wordt omgeschreven naar $c = \frac{Q}{m\Delta T}$. Hieruit wordt vervolgens c bepaald met de gemeten massa en temperatuurverschil.

# Resultaten


````python
# Hier de data en de analyse
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
data = np.loadtxt('Water.csv',delimiter=';')
time = data[:,0]
temp = data[:,1] + 273.15
V = 15 #V
I = 1.56 #A
P = V * I
def func(t,a,b):
    return a * t + b
values, covariance = curve_fit(func,time,temp)
x_test = np.linspace(-10,520,2000)
m = 0.5 #kg
delta_temp = np.max(temp)-np.min(temp)
Q = P * np.max(time)
# Q = m * c * delta_temp, -> c = Q / (m * delta_Temp)

c = Q / (m * delta_temp)
c_echt = 4186

plt.figure()
plt.plot(time,temp,'.',label='Metingen')
plt.plot(x_test,func(x_test,*values),'r--',label='Curve Fit')
plt.xlabel('$Time(s)$')
plt.ylabel('$Temp(K)$')
plt.xlim(-10,520)
plt.legend()
plt.grid()
plt.show()

print('De bepaalde soortelijke warmte is gelijk aan c = %.0f J/kgK.' %c)
print('De literatuurwaarde is gelijk aan c = %.0f J/kgK.' %c_echt)
foutmarge = (np.abs(c - c_echt) / c) * 100
print('Dat is een afwijking van %.0f %%.' %foutmarge)

````

# Discussie en conclusie

De gevonden waarde is $c = 4405 J/kgK$. Deze waarde heeft een afwijking van 5% ten opzichte van de literatuurwaarde van $c = 4186 J/kgK$. Dit is een relatief grote afwijking en kan veroorzaakt worden door een aantal dingen. Een belangrijke invloed is waarschijnlijk de gemeten massa van het water. Het experiment is namelijk een keer fout gedaan, waardoor de bak met water opnieuw gevuld moest worden. Deze is toen gevuld tot 500 mL, maar de massa van het water is niet precies bepaald. Het kan zijn dat de echte massa groter was, waardoor de bepaalde waarde lager uitkomt. Een andere oorzaak kan de efficiëntie van de verwarmingsbron zijn. Het kan bijvoorbeeld zo zijn dat deze in de realiteit met een afwijking van het gemeten vermogen werkt. Dit zou ook zorgen voor forse afwijking. Bovendien is het verstandiger om langer door te meten tot het water een hogere temperatuur heeft. Dit is voor een vervolgonderzoek aan te raden, evenals een precieze meting van de massa van het water doen, een groter vermogen voor het verwarmingselement gebruiken en onzekerheden van de meetinstrumenten te bepalen. Dit zou kunnen zorgen voor betere metingen, en een resultaat dat makkelijker te verantwoorden is door middel van een bepaalde onzekerheid.
